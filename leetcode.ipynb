```python
#两数之和
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        hashmap={} #hashmap is dictionary
        for idx,num in enumerate(nums):
            partner=target-num
            if partner in hashmap:
                return [idx,hashmap[partner]]
            hashmap[num]=idx
```

```python
#移动零
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        slow=0
        for fast in range(len(nums)):
            if nums[fast]!=0:
                nums[slow],  nums[fast]=nums[fast],nums[slow]
              
                slow+=1
        return nums
```

```python
128. 最长连续序列
if x - 1 in st: continue
意思：
如果 x 前面还有一个数，那 x 绝对不是连续序列的开头！
举个例子：
[1,2,3,4]
遍历到 2 → 2-1=1 存在 → 跳过
遍历到 3 → 3-1=2 存在 → 跳过
遍历到 4 → 4-1=3 存在 → 跳过
只有遍历到 1 → 1-1=0 不存在 → 这是起点！
continue =
本次循环剩下的代码不跑了，直接进入下一轮循环！
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        st=set(nums) #集合去重
        ans=0 #记录最长连续序列长度
        for x in st:
            if x-1 in st:#找到一个连续序列的开头了
                continue
            y=x+1
            while y in st:
                y+=1
            ans=max(ans,y-x)    
        return ans
```

```python
最多的水：
class Solution:
    def maxArea(self, height: List[int]) -> int:
        n=len(height)
        left=0
        right=n-1
        ans=0
        while left<right:
            sq=(right-left)*min(height[left],height[right])
            ans=max(ans,sq)
            if height[left]<height[right]:left+=1
            else:right-=1
        return ans
        
```

```python
三数之和：
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        n = len(nums)
        ans = []
        for i in range(n - 2):
            # 1. 固定第一个数，去重：跳过和前一个相同的数
            if i > 0 and nums[i] == nums[i - 1]:
                continue
            
            # 2. 双指针初始化
            j = i + 1
            k = n - 1
            
            while j < k:
                s = nums[i] + nums[j] + nums[k]
                if s < 0:
                    # 和小于0，左指针右移
                    j += 1
                elif s > 0:
                    # 和大于0，右指针左移
                    k -= 1
                else:
                    # 找到一组解，加入结果
                    ans.append([nums[i], nums[j], nums[k]])
                    j += 1
                    # 3. 对j去重：跳过所有和当前j相同的元素
                    while j < k and nums[j] == nums[j - 1]:
                        j=j+1

                    k -= 1
                    # 对k去重：跳过所有和当前k相同的元素
                    while j < k and nums[k] == nums[k + 1]:
                        k -= 1
                      
        
        return ans
```

```python


3. 无重复字符的最长子串

class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        ans = 0        # 记录最长无重复子串的长度
        left = 0       # 滑动窗口左边界
        cnt = defaultdict(int)  # 记录窗口内每个字符出现的次数

        # right 是滑动窗口右边界，c 是当前字符
        for right, c in enumerate(s):
            cnt[c] += 1  # 把当前字符加入窗口，次数+1

            # 如果当前字符出现次数 >1，说明窗口内有重复
            while cnt[c] > 1:
                cnt[s[left]] -= 1  # 把窗口左边的字符移出
                left += 1           # 左边界右移，缩小窗口

            # 现在窗口 [left, right] 一定无重复，更新最大值
            ans = max(ans, right - left + 1)

        return ans
```
